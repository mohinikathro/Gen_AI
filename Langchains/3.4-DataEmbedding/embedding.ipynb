##### Converting text into vector embeddings

```python
import os
from dotenv import load_dotenv
load_dotenv() # load all the env variables
```

```python
os.environ['OPENAI_API_KEY'] = os.getenv('OPENAI_API_KEY')
```

```python
from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(model= 'text-embedding-3-large')
embeddings
```

```python
text = 'this is a turotial on OPENAI embeddings'
query_res = embeddings.embed_query(text)
query_res
```

```python
query_res[0]
```

```python
len(query_res) # length of the embeddings for the provided text
```

```python
## Embeddings with custom vector length

embeddings_1024 = OpenAIEmbeddings(model= 'text-embedding-3-large', dimensions=1024)
res = embeddings_1024.embed_query(text)
len(res)
```

```python
res
```

```python
from langchain_community.document_loaders import TextLoader

doc = TextLoader('speech.txt').load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

txt_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 500,
    chunk_overlap = 50
)

final_txt = txt_splitter.split_documents(doc)
final_txt
```

```python
# Converting final_txt into vector embeddings and then storing it in a Vector DataBase
# ChromaDB for vectore store

from langchain_community.vectorstores import Chroma

db = Chroma.from_documents(final_txt, embeddings_1024)
```

```python
db
```

```python
## how to query from the vectorstore database using Similarity search

query = "It will be all the easier for us to conduct ourselves as belligerents"

retrieved_res = db.similarity_search(query)
print(retrieved_res)
```
