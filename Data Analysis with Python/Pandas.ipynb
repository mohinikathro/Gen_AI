```python
import pandas as pd
```

```python
# Series: is a 1D array like obj that can hold any data type,it is similar to a column in a table

data = [1,2,3,4,5]
series = pd.Series(data)
print('Series, \n', series)
print(type(series))
```

```python
# Create a Series from dictionary
# Key will become index, and values will be series values
data  = {'a':1, 'b':2, 'c':3}
series_dict = pd.Series(data)
print(series_dict)
```

```python
data = [10, 20, 30]
index = ['a', 'b', 'c']

pd.Series(data, index = index)
```

```python
## DataFrame: 2D 
## Create a DF from dict of list

data = {
    'Name': ['John', 'Jack', 'Daniel'],
    'Age': [25, 36, 28],
    'City': ['NJ', 'Chicago', 'NY']

}
df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(data)
```

```python
# Create a DF from list of dictionary, 

data = [
    {'Name': 'John', 'Age':36, 'City': 'NJ'},
    {'Name': 'John', 'Age':36, 'City': 'NJ'},
    {'Name': 'John', 'Age':36, 'City': 'NJ'},
    {'Name': 'John', 'Age':36, 'City': 'NJ'},
]

df = pd.DataFrame(data)
print(df)
print(type(df))
```
