```python
# Problem 1

# base class
class Animal:
    def sound(self):
        print('Animal makes sound')

# Derived class 1
class Dog(Animal):
    def sound(self):
        print('Dog barks')

# Derived class 2
class Cat(Animal):
    def sound(self):
        print('Cat meows')

# Polymorphism Function
def make_sound(animal):
    animal.sound()

# creating objects
dog1 = Dog()
cat1 = Cat()

make_sound(dog1)
make_sound(cat1)
```

```python
# Problem 2

# base class
class Shape():
    def area(self):
        pass

# derived class 1
class Square(Shape):
    def __init__(self, s):
        self.s = s

    def area(self):
        return self.s * self.s
    
# derived class 2
class Circle(Shape):
    def __init__(self, r):
        self.r = r

    def area(self):
        return 3.14 * self.r * self.r
    
# polymorphism function 
def print_area(shape):
    print(f'The area is: {shape.area()}')
    
# Creating objects
sq = Square(4)
cir = Circle(3)

print_area(sq)
print_area(cir)
```

```python
# Problem 3

class Vehicle:
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print('Car started')

class Bike(Vehicle):
    def start(self):
        print('Bike started')

def start_msg(vehicle):
    vehicle.start()

car1 = Car()
bike1 = Bike()

start_msg(car1)
start_msg(bike1)
```

```python
# Problem 4

class Employee:
    def get_salary(self):
        pass

class FulltimeEmp(Employee):
    def __init__(self, salary):
        self.salary = salary

    def get_salary(self):
        return self.salary
    
class ParttimeEmp(Employee):
    def __init__(self, hr_rate, hrs_worked):
        self.hr_rate = hr_rate
        self.hrs_worked = hrs_worked

    def get_salary(self):
        return self.hr_rate * self.hrs_worked

def print_salary(employee):
    print(f"The salary is: {employee.get_salary()}") 

femp = FulltimeEmp(45000)
temp = ParttimeEmp(40, 1600)

print_salary(femp)
print_salary(temp)
```

```python
# Problem 5
class Payment:
    def pay(self):
        pass

class Creditcard(Payment):
    def pay(self):
        print('Paid by Credit card')

class UPI(Payment):
    def pay(self):
        print('Paid by UPI')

class Cash(Payment):
    def pay(self):
        print('Paid by Cash')
        
def print_pay(payment):
    payment.pay()

pay1 = Creditcard()
pay2 = UPI()
pay3 = Cash()

print_pay(pay1)
print_pay(pay2)
print_pay(pay3)
```

```python
# Problem 6
class Notification:
    def send(self):
        pass

class EmailNoti(Notification):
    def send(self):
        print('Message sent by Email')

class SMSNoti(Notification):
    def send(self):
        print('Message sent by SMS')      
        
def print_noti(object):
    object.send()

email = EmailNoti()
sms = SMSNoti()

print_noti(email)
print_noti(sms)
```

```python
# Problem 7

class File:
    def read(self):
        pass

class TextFile(File):
    def read(self):
        print('Read text file')

class PDFFile(File):
    def read(self):
        print('Read PDF file')

def print_read(obj):
    obj.read()

txt1 = TextFile()
pdf1 = PDFFile()

print_read(txt1)
print_read(pdf1)
```

```python
# Problem 8

class Shape:
    def perimeter(self):
        pass

class Circle(Shape):
    def __init__(self, r):
        self.r = r 
    
    def perimeter(self):
        return 2 * 3.14 * self.r

class Rectangle(Shape):
    def __init__(self, l, w):
        self.l = l
        self.w = w

    def perimeter(self):
        return 2* (self.l + self.w)

def print_perimeter(obj):
    print(f'The primeter is: {obj.perimeter()}')

cir = Circle(4)
rect = Rectangle(3, 7)

print_perimeter(cir)
print_perimeter(rect)
```
