```python
# Problem 1

class Animal:

    def speak(self):
        print('Animal speaks')

class Dog(Animal):

    def speak(self):
        print("Dog barks")

animal1 = Animal()
animal1.speak()

dog1 = Dog()
dog1.speak()
```

```python
## Problem 2

# Creating super class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"My name is {self.name} and I am {self.age} years old.")

# Creating subclass
class Student(Person):  
    def __init__(self, name, age, grade):
        super().__init__(name, age) 
        self.grade = grade

    def introduce(self):
        super().introduce()
        print(f"I am in grade {self.grade}.")

stud1 = Person("Alex", 25)
stud1.introduce()  

stud2 = Student('Maya', 15, 6)
stud2.introduce()
```

```python
## Problem 3

class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def start(self):
        print('Vehicle started')

## Subclass

class Car(Vehicle):
    def __init__(self, brand, model):
        super().__init__(brand)
        self.model = model

    def start(self):
        print(f"Car {self.model} started")


## Another subclass
class ElectricCar(Car):
    def __init__(self, brand, model, battery_capacity):
        super().__init__(brand, model)
        self.battery_capacity = battery_capacity

    


## Creating an instance of ElectricCar 
car1 = ElectricCar('Toyota', 'Camry', 4000)
car1.start()

## Print all attribites 
print(f"Brand: {car1.brand}")
print(f'Model: {car1.model}')
print(f"Battery Capacity: {car1.battery_capacity} Wh")
```

```python
## Problem 4

class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def get_details(self):
        print(f"Employee {self.name} has {self.salary} salary")


# Subclass
class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department

    def get_details(self):
        return f"Employee {self.name} has {self.salary} salary and is from {self.department} department."

emp1 = Manager('Jeremy', "15K", 'IT')
emp1.get_details()
```

```python
## Problem 5

class Shape:
    def __init__(self, color):
        self.color = color

    def area(self):
        return 0

# Subclass
class Rectangle(Shape):
    def __init__(self, color, width, height):
        super().__init__(color) 
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

# Subclass
class Circle(Shape):
    def __init__(self, color, radius):
        super().__init__(color)    
        self.radius = radius

    def area(self):
        return 3.14 * (self.radius**2)
    
# Creating instance

rect = Rectangle('Black', 4, 5)
circle = Circle('Blue', 3)

print(f"Rectancle color: {rect.color}, area: {rect.area()}")
print(f"Circle color: {circle.color}, area: {circle.area()}")
```

```python
## Problem 7: Checking Inheritance

class Person:
    def __init__(self, name):
        self.name = name

class Employee(Person):
    def __init__(self, name, emp_id):
        super().__init__(name)
        self.emp_id = emp_id

def is_emp(obj):
        return isinstance(obj, Employee)
    
# Testing
p = Person('Alice')
e = Employee('Bob', 23131)
print(is_emp(p))
print(is_emp(e))
```
