### Inheritance (Single Inheritance: as only one child class)

```python
## Parent Class

class Car:
    def __init__(self, windows, doors, enginetype):   # constructor
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):  # method
        print(f"The person will drive the {self.enginetype} car.")
```

```python
car1 = Car(4, 4, 'petrol')
print(car1.drive())
```

```python
## Child 
class Tesla(Car):
    def __init__(self, windows, doors, enginetype, is_selfdriving):
        super().__init__(windows, doors, enginetype) # call the parent class, and calling the init method from parent class
        self.is_selfdriving = is_selfdriving

    def selfdriving(self):
        print(f"Tesla supports self driving: {self.is_selfdriving}")
```

```python
tesla1 = Tesla(4, 4, 'Electric', True)
tesla1.selfdriving() # defined in child class 
tesla1.drive() # defined in parent class
```

### Multiple Inheritance: when child class inheritance from more than 1 class

```python
class Animal: # base class one
    def __init__(self, name):
        self.name = name

    def speak(self):
        print('Subclass must implement this method')


# Base class 2
class Pet:
    def __init__(self, owner):
        self.owner = owner
    
## Derived Class
class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)

    def speak(self):
        return f"{self.name} says woof"
    
## Create an object
dog1 = Dog('Buddy', 'Alex')
dog1.speak()
```
