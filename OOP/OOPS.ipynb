```python
## Define a class with instance methods

class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        print(f"{self.name} says woof")

dog3 = Dog('Alex', 3)
dog3.bark()

dog4 = Dog('lucy', 5)
dog4.bark()
```

```python
## Modeling a bank acocunt

# Define a class for bank acocunt
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount): # instance method
        self.balance += amount # attribute
        print(f"{amount} is deposited. New balance is {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print('Insufficient balance')
        
        else:
            self.balance -= amount
            print(f"{amount} is withdrawn. New balance is {self.balance}")

    def get_balance(self):
        return self.balance
                 

## Creating an account

account = BankAccount('Alex', 5000)
print(account.balance)

# calling instance method to deposit 100 in account 
account.deposit(100)

# withdraw money
account.withdraw(300)

# full balance
account.get_balance()

# withdraw amount more than the actual balance
account.withdraw(4900)
```

```python
## Creating a class

class Human:
    def __init__(self, name, gender, occupation): # method
        self.name = name # property 1
        self.gender = gender # property 2
        self.occupation = occupation # property 3

    def do_work(self):  # method 1 
        if self.occupation == 'Tennis player':
            print(self.name, 'plays tennis')
        elif self.occupation == 'actor':
            print(self.name, 'shoots films')
    
    def speaks(self): # method 2
        print(self.name, 'says how are you?')

    def bio(self):
        print(f"{self.name} is a {self.gender} and their occupation is {self.occupation}")

tom = Human('Tom Cruise', 'male', 'actor') # insatnce 1
tom.do_work()
tom.speaks()
tom.bio()

maria = Human('Maria Sharapova', 'female', 'Tennis player') # insatnce 2 
maria.bio()
maria.do_work()
maria.speaks()
maria.gender
```
