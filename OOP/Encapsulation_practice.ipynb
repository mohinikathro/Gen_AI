```python
# Problem 2
class Account:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance +=  amount
        else: 
            print('Deposit amount must be positive')
        
    def withdraw(self, amount):
        if amount <= self.__balance and amount >0:
            self.__balance -= amount
        else:
            print('Insufficient balance')

    def get_balance(self):
        return self.__balance
    
acc1 = Account(500)

acc1.deposit(20)
print(acc1.get_balance())

acc1.withdraw(50)
print(acc1.get_balance())

acc1.withdraw(1000)
```

```python
# Problem 3
class Student:
    def __init__(self, name, marks):
        self.__name = name
        self.__marks = marks

    def get_marks(self):
        return self.__marks
    
    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print('The marks must be between 0 and 100')

stud1= Student('Axel', 75)

print(stud1.get_marks())

stud1.set_marks(123)

stud1.set_marks(0)
print(stud1.get_marks())
        
```

```python
# Problem 4
class Product:
    def __init__(self, name, price):
        self.__name = name
        self.__price = price

    def get_name(self):
        return self.__name
    
    def get_price(self):
        return self.__price
    
    def set_price(self, price):
        if price >0:
            self.__price = price
        else:
            print('Price is invalid.')

    def apply_discount(self, percentage):
        if 0 < percentage <=100:
            self.__price = round(self.__price * (1 - percentage /100), 2)
        else:
            print('Invalid discount percentage')


product1 = Product('Chips', 4.99)

print(product1.get_price())

product1.set_price(0)

product1.set_price(6.99)
print(product1.get_price())

product1.apply_discount(100)
print(product1.get_price())

product1.set_price(6.99)
print(product1.get_price())

product1.apply_discount(10)
print(product1.get_price())

```

```python
# Problem 5

class LoginSystem:
    def __init__(self, username, password):
        self.__username = username
        self.__password = password
        self.__attempts = 0


    def login(self, password):
        if self.__attempts >= 3:
            print('Account locked')
            return
        
        if password == self.__password:
            print('Login Successful')
            self.__attempts = 0 # reset after success
        else:
            self.__attempts += 1
            print('Incorrect password')

            if self.__attempts >=3:
                print('Account locked')

user = LoginSystem('Alex', 'secure123')

user.login('wrong')
user.login('1234')
user.login('3bwdi')
user.login('secure123')
            
        

        
```
