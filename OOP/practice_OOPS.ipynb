```python
## Problem 1

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi, my name is {self.name} and I am {self.age} years old.")

person1 = Person('Alex', 5)
person1.introduce()


## Problem 2: class with default value
class Student:
    def __init__(self, name, marks=0):
        self.name = name
        self.marks = marks

    def is_passed(self):
        if self.marks >= 40:
            return True
        else:
            return False
        
print('\n')
student1 = Student('Mark', 40)
print(student1.is_passed())

student2 = Student('Elsa')
print(student2.is_passed())


## Problem 3: Bank Account
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount): # instance method
        self.balance += amount # attribute
        print(f"{amount} is deposited. New balance is {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print(f'Insufficient balance, can not withdraw {amount} as account balance is {self.balance}')
        
        else:
            self.balance -= amount
            print(f"{amount} is withdrawn. New balance is {self.balance}")

    def get_balance(self):
        return self.balance
                 

## Creating an account

print('\n')
account = BankAccount('Alex', 5000)
print(account.balance)

# calling instance method to deposit 100 in account 
account.deposit(100)

# withdraw money
account.withdraw(300)

# full balance
print(account.get_balance())

# withdraw amount more than the actual balance
account.withdraw(4900)


## Problem 4: 
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def get_details(self):
        print(f"The car is a {self.brand} {self.model}, model from year {self.year}.")

car1 = Car('Toyota', 'Camry', 2016 ) # Object 1
car2 = Car('Honda', 'Civic', 2009)  # Object 2
car3 = Car('Ford', 'Mustang', 2000) # Object 3

print('\n')
car1.get_details()
car2.get_details()
car3.get_details()



## Problem 5: using __str__
class Book():
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def __str__(self):
        return f"{self.title} by {self.author} - ${self.price}"


book1 = Book('Harry Potter', 'J.K Rowling', 120)
print('\n', book1)
```

```python
## Problem 6: class with list attributes

class ShoppingCart():
    def __init__(self): # Attribute with empty list 
        self.items = []

    def add_item(self, item):
        self.items.append(item)

    def remove_item(self, item):
        if item in self.items:
            self.items.remove(item)
            print(f"Removed {item} from cart")
        else:
            print(f'{item} not found in cart')

    def show_items(self):
        if not self.items:
            print('Shopping cart is empty')
        else:
            for item in self.items:
                print(item)

cart1 = ShoppingCart()
print('\n')
cart1.add_item('Apple')
cart1.add_item('Banana')
cart1.add_item('Cheery')


cart1.remove_item('Guava')
cart1.remove_item('Apple')
cart1.show_items()
```
