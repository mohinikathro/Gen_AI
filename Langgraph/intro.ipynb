```python
from pydantic import BaseModel
```

```python
from dataclasses import dataclass

@dataclass
class Person1():
    name: str
    age: int
    city: str

person = Person1(name='Axel', age=25, city='California')
print(person)
```

```python
# Do not get error here for wrong value in city which should be string
person1 = Person1(name='Sue', age=15, city=12)
print(person1)
```

```python
class Person(BaseModel):
    name: str
    age: int
    city: str

person = Person(name='Axel', age=25, city='California')
print(person)
```

### 2. model with Optional fields

#### Add optional fields using Python's Optional type

```python
from typing import Optional
class Employee(BaseModel):
    id: int
    name: str
    department: str
    salary: Optional[float] = None # Optional with default value None
    is_active: Optional[bool] = True # Optional with default value None
```

```python
emp1 = Employee(id=1, name='John', department='IT')
print(emp1) # salary and is_active values are not given 
```

```python
emp2 = Employee(id=1, name='Jane', department='HR', salary=60000, is_active=False)
print(emp2)
```

```python
from typing import List

class Classroom(BaseModel):
    room_no: str
    students: List[str] # list of strings
    capacity: int
```

```python
room1 = Classroom(
    room_no='A101',
    students= ['Alice', 'Bob', 'Charlie'],
    capacity= 30
)

print(room1)
```

```python
room1 = Classroom(
    room_no='A101',
    students= ('Alice', 'Bob', 'Charlie'),
    capacity= 30
)

print(room1)
```

```python
try:
    invalid_val = Classroom(
        room_no='A1',
        students=['Dane', 123],
        capacity= 25
    )
except ValueError as e:
    print(e)

```

### Nested Models


#### Creating complex structures with nested models:

```python
class Address(BaseModel):
    street: str
    city: str
    zip_code: int

class Customer(BaseModel):
    customer_id: int
    name: str
    address: Address

## Creating a customer with nested address
cust1 = Customer(
    customer_id=1, 
    name= 'Esha',
    address= {'street': '123 Main St', 'city': 'Boston', 'zip_code': '02108'}
)

print(cust1)
```

### Pydantics Fields: Customization and Contraints

```python
from pydantic import Field

class Item(BaseModel):
    name: str = Field(min_length=2, max_length=50)
    price: float = Field(gt=0, le=1000)
    quantity: int = Field(ge=0)


# Validate instance
item = Item(name='Book', price=100, quantity=10)
print(item)
```

```python
# Validate instance
item = Item(name='Book', price=100, quantity=10)
print(item)
```
