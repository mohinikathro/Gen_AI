```python
# Function copy

def Welcome():
    return "Welcome to the python class."

Welcome()
```

```python
wel = Welcome # one function is copied to another function

print(wel())

# delete welcome method
del Welcome 

print(wel()) # even after deleting the reference function, the function copy is still there
```

```python
## Things defined outside the sun_method, can be accessed

def main_welcome(msg):
    
    def sub_welcome_method():
        print('Welcome to advance python course')
        print(msg)
        print('Please learn this concepts properly')

    return sub_welcome_method()

main_welcome('Welcome everyone')
```

```python
def main_welcome(func):
    
    def sub_welcome_method():
        print('Welcome to advance python course')
        func('Welcome everyone to this tutorial')
        print('Please learn this concepts properly')

    return sub_welcome_method()

main_welcome(print)
```
