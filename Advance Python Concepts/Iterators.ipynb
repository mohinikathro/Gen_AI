```python
# String iterator

my_string = 'Hello'
string_iterator = iter(my_string)
```

```python
print(next(string_iterator))
print(next(string_iterator))
```

```python
## Exception handling

try:
    print(next(string_iterator))
except StopIteration:
    print('There are no more elements left in the string')
```
