#### Building a simple LLM Application using LCEL (LangChain Expression Language)

```python
### Open Source models: Llama3, Gemma2, Mistral (using these models using Groq)
```

```python
import os
from dotenv import load_dotenv
load_dotenv()

import openai
openai.api_keys = os.getenv('OPENAI_API_KEY')

groq_api_key = os.getenv('GROQ_API_KEY')
```

```python
from langchain_groq import ChatGroq
from langchain_openai import ChatOpenAI

model = ChatGroq(model = 'llama-3.1-8b-instant', groq_api_key = groq_api_key)
model
```

```python
from langchain_core.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage(content='Transalte the following from English to French'),
    HumanMessage(content='Hello, How are you?')
]

res = model.invoke(messages)
```

```python
res.content
```

```python
res
```

```python
res.response_metadata
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()

parser.invoke(res)
```

```python
### Using LCEL- chaining the components

chain = model | parser

chain.invoke(messages)
```

```python
### Prompt Templates

from langchain_core.prompts import ChatPromptTemplate

generic_template = 'Translate the following into {language}'

prompt = ChatPromptTemplate.from_messages(
    [('system', generic_template), ('user', '{text}')]
)
```

```python
result = prompt.invoke({'language': 'French', 'text':'Hello'})
```

```python
result.to_messages()
```

```python
## Chaining together components with LCEL

chain = prompt | model | parser

chain.invoke({'language':'French', 'text':'Hello'})
```
